```python
def generate_basic_winning_states():
  NUM_WINNING_STATES = 10
  winning_states = [{"field": np.zeros(5, dtype=np.int8), "hand": np.ones((5), dtype=np.int8), "discard": np.zeros((5), dtype=np.int8), "deck": []} for i in range(NUM_WINNING_STATES)]

  for i in range(NUM_WINNING_STATES//2):
    winning_states[i]["field"][i] = 4

  for i in range(NUM_WINNING_STATES//2, NUM_WINNING_STATES):
    winning_states[i]["field"] = np.ones(5, np.int8)
    winning_states[i]["field"][i-5] = 0

  return winning_states
```

```python
winning_states = generate_basic_winning_states()
print(winning_states)
```

```python
env.reset()
print(env.game._get_board())
for field in env.game.players[0]:
  env.game.players[0][field] = np.copy(winning_states[4][field])
print(env.game._get_board())
```
